```python
# Import Library yang dibutuhkan
!pip install youtube-comment-downloader
from youtube_comment_downloader import YoutubeCommentDownloader
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import re
```

```python
# Scraping data komen youtube

# Input link video youtube
video_url = "https://www.youtube.com/watch?v=lmcwEO9Lqe8"

# Ambil data comment video youtube
downloader = YoutubeCommentDownloader()

comments = []

for comment in downloader.get_comments_from_url(video_url):

    comments.append({
        'comment': comment['text'],
        'likes': comment['votes'],
        'time': comment['time'],
        'author': comment['author']
    })

    # limit komentar
    if len(comments) >= 1000:
        break

# Menyimpan File hasil scrapping
df = pd.DataFrame(comments)
filename = "analisis_sentimen.csv"
df.to_csv(filename, index=False, encoding='utf-8')

print(f"Proses scraping selesai! Berhasil menyimpan {len(df)} komentar ke dalam file: '{filename}'")
```

### 2. Langkah-Langkah Data Cleaning

- menghapus label "(edited)"

#### 2.5 Normalisasi Teks (`cleaning_text`)

Fungsi `cleaning_text()` menjalankan pipeline pembersihan teks secara
berurutan:

| Urutan | Proses | Contoh |
|---|---|---|
| 1 | **Lowercase** | `"MBG Korupsi"` → `"mbg korupsi"` |
| 2 | **Hapus URL** | `"https://t.co/xyz"` → `""` |
| 3 | **Hapus Mention** | `"@user123"` → `""` |
| 4 | **Hapus Hashtag** | `"#MBG"` → `""` |
| 5 | **Normalisasi Satuan Angka** | `"300jt"` → `"300 juta"`, `"1,2T"` → `"1,2 triliun"` |
| 6 | **Hapus Karakter Berulang** | `"bagussss"` → `"baguss"` |
| 7 | **Hapus Karakter Khusus** | Hanya huruf, angka, spasi, `,` `.` `%` yang dipertahankan |
| 8 | **Normalisasi Spasi** | Spasi ganda & newline → 1 spasi |
| 9 | **Normalisasi Slang** | `"gak"` → `"tidak"`, `"yg"` → `"yang"` |

#### 2.6 Normalisasi Kamus Slang (`SLANG_DICT`)
Kamus normalisasi mencakup 3 kategori:
- **Singkatan umum** — `"yg"`, `"krn"`, `"tdk"`, `"gak"`, dll.
- **Akronim domain** — `"sppg"`, `"bgn"`, `"apbn"`, `"dpr"`, dll.
- **Typo konsisten** — `"negatip"` → `"negatif"`, `"trilyun"` → `"triliun"`, dll.

#### 2.7 Filter data dengan jumlah kata >= 5

---

### 3. Output Setelah Cleaning

- Kolom baru **`clean_text`** ditambahkan ke DataFrame sebagai hasil cleaning
- Kolom `comment` (teks asli) tetap dipertahankan untuk keperluan validasi
- Data siap digunakan untuk proses selanjutnya:
  tokenisasi, vektorisasi, dan pemodelan sentimen

---

```python
# Data Cleaning & Data Quality Check

# Normalisasi kata informal / slank
SLANG_DICT = {
    # Singkatan umum
    "yg": "yang", "dgn": "dengan", "dg": "dengan", "dr": "dari",
    "krn": "karena", "krna": "karena", "krena": "karena",
    "utk": "untuk", "tuk": "untuk", "buat": "untuk",
    "sdh": "sudah", "udh": "sudah", "udah": "sudah",
    "blm": "belum", "blum": "belum",
    "lg": "lagi", "lagi": "lagi",
    "jg": "juga", "juga": "juga",
    "tp": "tapi", "tapi": "tapi",
    "klo": "kalau", "klu": "kalau", "klau": "kalau", "kalo": "kalau",
    "gak": "tidak", "ga": "tidak", "gk": "tidak", "nggak": "tidak",
    "enggak": "tidak", "ngga": "tidak", "tdk": "tidak", "tak": "tidak",
    "gw": "saya", "gue": "saya", "aku": "saya",
    "lo": "kamu", "lu": "kamu", "loe": "kamu",
    "bpk": "bapak", "pak": "bapak", "ibu": "ibu",
    "sm": "sama", "ama": "sama",
    "skrg": "sekarang", "skrang": "sekarang",
    "bs": "bisa", "bsa": "bisa",
    "gmn": "gimana",
    "msh": "masih",
    "kmn": "kemana", "kmana": "kemana",
    "ntr": "nanti", "ntar": "nanti",
    "nih": "ini", "ni": "ini",
    "tuh": "itu", "tu": "itu",
    "emg": "memang", "emang": "memang",
    "mnrt": "menurut",
    "dpt": "dapat", "dpat": "dapat",
    "hrs": "harus",
    "lbh": "lebih",
    "bgs": "bagus",
    "jk": "jika",
    "ttg": "tentang",
    "jd": "jadi", "jdi": "jadi",
    "pd": "pada",
    "dl": "dulu", "dlu": "dulu",
    "bnyk": "banyak", "byk": "banyak",
    "trs": "terus", "truss": "terus",
    "org": "orang", "orng": "orang",
    "pst": "pasti",
    "pemerintah": "pemerintah",
    "sppg": "satuan pelayanan pemenuhan gizi",
    "bgn": "badan gizi nasional",
    "apbn": "anggaran pendapatan belanja negara",
    "umkm": "usaha mikro kecil menengah",
    "sdm": "sumber daya manusia",
    "dpr": "dewan perwakilan rakyat",
    # Typo / penulisan tidak konsisten
    "negatip": "negatif", "positip": "positif",
    "kwalitas": "kualitas", "kwaliatas": "kualitas",
    "aktifitas": "aktivitas",
    "praktek": "praktik",
    "nabi": "nabi",
    "triliyun": "triliun", "trilyun": "triliun",
}

# Proses cleaning data
def cleaning_text(text: str, normalize_slang: bool = True):
    # Lowercase semua huruf
    text = text.lower()

    # Hapus URL (http, https, www)
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Hapus mention YouTube/sosmed (@username)
    text = re.sub(r"@[\w\-\.]+", " ", text)

    # Hapus hashtag
    text = re.sub(r"#\w+", " ", text)

    # 6. Normalisasi satuan angka informal
    text = re.sub(r"(\d+)\s*t\b", r"\1 triliun", text)  # 335t  → 335 triliun
    text = re.sub(r"(\d+)\s*m\b", r"\1 miliar", text)   # 10m   → 10 miliar
    text = re.sub(r"(\d+)\s*jt\b", r"\1 juta", text)    # 300jt → 300 juta
    text = re.sub(r"(\d+)\s*rb\b", r"\1 ribu", text)    # 15rb  → 15 ribu
    text = re.sub(r"(\d+)\s*k\b", r"\1 ribu", text)     # 10k   → 10 ribu

    # 7. Hapus karakter berulang berlebihan (maks 2)
    text = re.sub(r"([^.])\1{2,}", r"\1\1", text)

    # 8. Hapus semua karakter selain huruf, angka, dan spasi
    text = re.sub(r"[^a-z0-9\s,\.%]", " ", text)

    # 9. Normalisasi spasi berlebih & whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # 10. Normalisasi slang & singkatan Bahasa Indonesia
    if normalize_slang:
        words = text.split()
        words = [SLANG_DICT.get(word, word) for word in words]
        text = " ".join(words)

    return text


df['clean_text'] = df['comment'].apply(cleaning_text)

# Filter data dengan jumlah kata >=5
df['panjang_teks']   = df['clean_text'].apply(len)
df['jumlah_kata']        = df['clean_text'].apply(lambda x: len(x.split()))

df = df[df['jumlah_kata'] >= 5]

# Check Tipe data
print(f"Tipe data pada setiap kolom dataset \n{df.dtypes}\n")

# Check Missing Values
print(f"Jumlah missing value pada dataset \n{df.isnull().sum()}\n")

# Handle Duplicate Values
df = df.drop_duplicates()
print(f"Jumlah data duplikat pada dataset : {df.duplicated().sum()}\n")

# Cek jumlah baris
print(f"Jumlah baris pada dataset: {len(df)}")

# Ubah tipe data kolom like menjadi integer
df['likes'] = df['likes'].astype('int')

# Normalisasi kolom time
# Hapus label "(edited)"
df['time'] = df['time'].str.replace(r'\(edited\)', '', regex=True).str.strip()
```

#### 6.4.2 Word Cloud

Tujuan : Untuk mengetahui kata/keyword apa yang paling sering muncul dalam diskusi MBG

| Proses | Detail |
|---|---|
| **Case Folding** | Mengubah seluruh teks menjadi huruf kecil (lowercase) untuk menghindari duplikasi kata akibat perbedaan kapitalisasi. |
| **Stopword Removal** | Menghapus kata-kata umum yang tidak memiliki makna penting dalam analisis, seperti "dan", "yang", "akan", serta stopword tambahan yang sering muncul pada komentar media sosial. |
| **Stemming** | Mengubah kata berimbuhan menjadi bentuk dasar menggunakan library Sastrawi |
| **Penggabungan Teks** | Menggabungkan seluruh komentar yang telah dibersihkan menjadi satu kumpulan teks untuk menghitung frekuensi kemunculan kata. |
| **Word Cloud Generation** | Membuat visualisasi Word Cloud menggunakan library WordCloud berdasarkan frekuensi kemunculan kata dalam dataset. |


```python
# Install library
!pip install Sastrawi

# Import library
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from wordcloud import WordCloud

# Stopwords
stop_factory = StopWordRemoverFactory()
stop_words = set(stop_factory.get_stop_words())

custom_stopwords = {
    'aja',
    'nya',
    'sih',
    'nih',
    'dong',
    'ya',
    'lah',
    'kan',
    'jadi',
    'kalau',
    'sudah',
    'akan',
    'juga',
    'saya',
    'kami',
    'kita',
    'mereka'
}

stop_words.update(custom_stopwords)


# Stemmer
stem_factory = StemmerFactory()
stemmer = stem_factory.create_stemmer()

# Function Cleaning Topic
def clean_topic(text):

    text = str(text).lower()

    words = text.split()

    # Remove stopwords
    words = [
        word for word in words
        if word not in stop_words
    ]

    # Stemming
    words = [
        stemmer.stem(word)
        for word in words
    ]

    return " ".join(words)


# Apply ke dataframe
df['clean_topic'] = df['clean_text'].apply(clean_topic)

# Gabungkan semua komentar
text = ' '.join(df['clean_topic'].dropna())

# Generate WordCloud
wordcloud = WordCloud(
    width=1200,
    height=600,
    background_color='white',
    max_words=100
).generate(text)

# Visualisasi
plt.figure(figsize=(14,7))

plt.imshow(wordcloud, interpolation='bilinear')

plt.axis('off')

plt.title(
    'Word Cloud Komentar Program MBG',
    fontsize=16,
    fontweight='bold'
)

plt.show()
```

- Happy (19.1%) : menunjukkan bahwa masih terdapat kelompok masyarakat yang mendukung program karena melihat manfaatnya bagi pemenuhan gizi anak. Contoh kalimat : "terima kasih pemerintah atas program makan bergizi gratis ini. sebagai orang tua, kami sangat terbantu karena gizi anak di sekolah jadi lebih terjamin dan meringankan beban pengeluaran harian. anak anak jadi lebih semangat belajar dan konsentrasinya meningkat."

- Love (0.1%) : hampir tidak muncul, menunjukkan bahwa dukungan yang ada cenderung bersifat rasional dan pragmatis, bukan dukungan emosional yang sangat kuat.

**Keterbatasan**

Beberapa komentar menunjukkan ketidaksesuaian antara hasil prediksi model emosi dengan konteks aktual kalimat. Sebagai contoh, komentar yang mengandung kritik keras terhadap tokoh politik pada beberapa kasus diklasifikasikan sebagai emosi love. Hal ini menunjukkan keterbatasan model pre-trained dalam memahami konteks politik, sarkasme, dan opini kompleks yang sering ditemukan pada komentar media sosial. Oleh karena itu, hasil emotion analysis perlu diinterpretasikan secara hati-hati dan sebaiknya digunakan sebagai pelengkap analisis sentimen, bukan sebagai satu-satunya dasar pengambilan kesimpulan.
